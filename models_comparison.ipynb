# Comparison between **Eyes Open (EO)** and **Eyes Closed (EC)**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os
```

#### Gaussian Naive Bayes

```python
if not os.path.exists('images'):
    os.makedirs('images')

df_results = pd.read_csv('eeg_model_results_eo_ec.csv')

chosen_model = 'Gaussian Naive Bayes' 
df_model = df_results[df_results['Model'] == chosen_model]

df_melted = df_model.melt(id_vars=['Condition'], 
                          value_vars=['Accuracy', 'AUC_ROC', 'F1_Score'], 
                          var_name='Metric', 
                          value_name='Score')
              
plt.figure(figsize=(10, 6))
ax = sns.barplot(data=df_melted, x='Metric', y='Score', hue='Condition', palette='muted')

ax.set_ylim(0.20, 0.98)
ax.set_yticks(np.arange(0.20, 0.98, 0.05))

for container in ax.containers:
    ax.bar_label(container, fmt='%.3f', padding=3, fontsize=11)

plt.title(f'Performance by Metric: {chosen_model}', fontsize=15, fontweight='bold')
plt.ylabel('Score')
plt.xlabel('Metric')
plt.legend(title='Condition', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()

file_name = f"performance_{chosen_model.replace(' ', '_').lower()}.png"
plt.savefig(f'images/{file_name}', dpi=300, bbox_inches='tight')

plt.show()
```
